```python
##create a simple grid-based spatial analysis
```

```python
# Create geographic grid cells

gdf["lat_grid"] = gdf["Latitude"].round(3)
gdf["lon_grid"] = gdf["Longitude"].round(3)

# Count street lights in each grid cell
grid_counts = (
    gdf.groupby(["lat_grid", "lon_grid"])
    .size()
    .reset_index(name="street_light_count")
)

grid_counts.head(10)
```

```python
### 1,017 records share the same exact Geo Point, showing that multiple street-light assets can exist at one geographic location.
### .round(3) groups nearby latitude and longitude values into small geographic grid cells, allowing us to count street lights within each area.
```

```python
# Find grid cells with the most street lights

grid_counts.sort_values(
    by="street_light_count",
    ascending=False
).head(10)
```

```python
import matplotlib.pyplot as plt
```

```python
# Visualize street-light concentration by grid cell

plt.figure(figsize=(10, 8))

plt.scatter(
    grid_counts["lon_grid"],
    grid_counts["lat_grid"],
    s=grid_counts["street_light_count"] / 2,
    c=grid_counts["street_light_count"],
    alpha=0.7
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Street Light Concentration by Grid Cell")
plt.colorbar(label="Number of Street Lights")

plt.show()
```

```python
# Finding
# The grid-based spatial analysis shows that street lights are not evenly distributed across the study area. 
#Some geographic grid cells contain substantially higher concentrations of street-light assets than others. 
#This analysis helps identify areas with a high density of street-light assets.
```

```python
# Summary statistics of street-light counts per grid cell

grid_counts["street_light_count"].describe()
```

```python
# Finding
# There are 304 geographic grid cells in the study area.
# The average grid cell contains approximately 349 street lights,
# while the median contains 292 street lights.
# Street-light counts range from 2 to 1,827 per grid cell,
# indicating substantial variation in street-light concentration.
```

```python
# Identify grid cells with above-average street-light concentration

average_count = grid_counts["street_light_count"].mean()

high_density_cells = grid_counts[
    grid_counts["street_light_count"] > average_count
]

print("Average street-light count:", average_count)
print("Number of above-average grid cells:", len(high_density_cells))

high_density_cells.head(10)
```

```python
# Display the 10 highest-density grid cells

top_density_cells = (
    grid_counts
    .sort_values("street_light_count", ascending=False)
    .head(10)
)

top_density_cells
```

```python
# Visualize the 10 highest-density grid cells

plt.figure(figsize=(10, 8))

plt.scatter(
    top_density_cells["lon_grid"],
    top_density_cells["lat_grid"],
    s=top_density_cells["street_light_count"] / 3,
    c=top_density_cells["street_light_count"],
    cmap="viridis",
    alpha=0.8
)

plt.colorbar(label="Number of Street Lights")

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Top 10 Highest-Density Street-Light Grid Cells")

plt.show()
```

```python
# Finding
# The spatial visualisation highlights the 10 grid cells with the highest
# concentrations of street-light assets.
# The highest-density grid cell contains 1,827 street-light records,
# followed by cells containing 1,770 and 1,386 records.
# These locations represent areas with particularly high concentrations
# of street-light assets within the study area.
```

```python
## spatial relationship between label and concentration
```

```python
# Calculate average label value for each grid cell

grid_label_analysis = (
    gdf.groupby(["lat_grid", "lon_grid"])
    .agg(
        street_light_count=("label", "size"),
        average_label=("label", "mean")
    )
    .reset_index()
)

grid_label_analysis.head(10)
```

```python
# Calculate correlation between street-light concentration and average label

correlation = grid_label_analysis[
    ["street_light_count", "average_label"]
].corr()

correlation
```

```python
# Finding
# The correlation between street-light concentration and average label
# is approximately -0.009.
# This indicates a negligible linear relationship between the two variables.
# Therefore, areas with more street-light records do not necessarily
# have higher or lower average label values.
```

```python
# Visualize the relationship between street-light concentration and average label

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    grid_label_analysis["street_light_count"],
    grid_label_analysis["average_label"],
    alpha=0.6
)

plt.xlabel("Number of Street Lights")
plt.ylabel("Average Label")
plt.title("Street-Light Concentration vs Average Label")

plt.show()
```

```python
# Finding
# The scatter plot shows no clear linear pattern between the number of
# street lights and average label value across grid cells.
# This is consistent with the correlation coefficient (r = -0.009),
# indicating a negligible linear relationship between the variables.
```

```python
# Overall Finding
# The spatial analysis shows that street-light assets are unevenly
# distributed across the study area, with some grid cells containing
# substantially higher concentrations than others.
# The highest-density grid cell contains 1,827 street-light records.
# However, the correlation analysis (r = -0.009) indicates a negligible
# linear relationship between street-light concentration and average
# label value. Therefore, areas with high concentrations of street lights
# do not necessarily have higher or lower average label values.
```

```python
# ## SQL-Based Asset Analysis

# The processed street-light dataset is stored in a SQL database to demonstrate structured querying and asset data management. 
#SQL is used to retrieve, filter, aggregate, and analyse street-light records based on attributes such as asset ID, geographic coordinates, 
#and label values.
# The SQL analysis will provide summary information about the street-light assets and support further reporting and visualisation in Power BI.
```

```python
import sqlite3

# Create SQLite database
conn = sqlite3.connect("street_lights.db")

# Export street-light data to SQL
gdf.drop(columns="geometry").to_sql(
    "street_lights",
    conn,
    if_exists="replace",
    index=False
)

print("Street-light data exported to SQL successfully.")
```

```python
# Check the number of records in the SQL table

query = """
SELECT COUNT(*) AS total_street_lights
FROM street_lights;
"""

pd.read_sql_query(query, conn)
```

```python
# Retrieve the first 10 street-light records

query = """
SELECT *
FROM street_lights
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# Count street lights by label category

query = """
SELECT 
    label_category,
    COUNT(*) AS street_light_count
FROM street_lights
GROUP BY label_category
ORDER BY street_light_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Find grid cells with the highest number of street lights

query = """
SELECT 
    lat_grid,
    lon_grid,
    COUNT(*) AS street_light_count
FROM street_lights
GROUP BY lat_grid, lon_grid
ORDER BY street_light_count DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```
